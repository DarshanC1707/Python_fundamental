## Module 4 – Logging

### 1. What is Logging?

Definition

Logging is the process of recording information about what is happening inside a program.

Real-Life Example

In a shop, a security camera records what happens throughout the day. Similarly, logging records what happens inside a program.

Example

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Program started")
logging.warning("Low storage")
```

Explanation: Logging records important events that happen while the program runs.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("User logged in")
logging.error("Database connection failed")
```

Explanation: Logs help developers understand normal activities and errors.

### 3. print() vs Logging

Definition

print() displays information on the screen, while logging records program events with different levels and can save them to files.

Real-Life Example

print() is like telling someone something verbally. Logging is like writing the information in a record book for future reference.

Example

```python
import logging

print("User logged in")

logging.basicConfig(level=logging.INFO)
logging.info("User logged in")
```

Explanation: print() is mainly for simple display, while logging is better for monitoring and debugging applications.

### 4. Log Levels

Definition

Log levels indicate the importance or severity of a log message.

Python provides five common levels:

DEBUG – Detailed information for debugging

INFO – Normal program activity

WARNING – Something may cause a problem

ERROR – A serious problem occurred

CRITICAL – A very serious problem that may stop the program

Real-Life Example

Think of a hospital:

DEBUG → Doctor checking detailed information

INFO → Patient admitted

WARNING → Patient needs attention

ERROR → Treatment problem

CRITICAL → Life-threatening emergency

Example

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debug information")
logging.info("Program started")
logging.warning("Low memory")
logging.error("File not found")
logging.critical("System failure")
```

Explanation: Different levels help us understand how serious an event is.

### 5. Log Formatters

Definition

A Log Formatter defines how log messages should look when they are displayed or saved.

Real-Life Example

A school record may contain date, student name, and marks in a fixed format. Similarly, a log formatter defines the structure of a log message.

Example

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s - %(message)s"
)

logging.info("Program started")
logging.warning("Low memory")
```

Explanation: The formatter controls the appearance of each log message.

### 6. File Handlers

Definition

A File Handler sends log messages to a file instead of only displaying them on the screen.

Real-Life Example

A company writes important activities in a register so they can check them later. A file handler does the same by saving logs in a file.

Example

```python
import logging

logging.basicConfig(
    filename="application.log",
    level=logging.INFO
)

logging.info("Program started")
logging.error("Something went wrong")
```

Explanation: The logs are stored inside application.log for future reference.

### 7. Console Handlers

Definition

A Console Handler sends log messages directly to the console or terminal.

Real-Life Example

A teacher announces information directly to students. Similarly, a console handler displays logs directly in the terminal.

Example

```python
import logging

logger = logging.getLogger("MyApp")
logger.setLevel(logging.INFO)

console_handler = logging.StreamHandler()
logger.addHandler(console_handler)

logger.info("Program started")
```

Explanation: StreamHandler() displays log messages in the console.

### 8. Rotating Logs

Definition

Rotating logs means automatically creating a new log file when the current log file reaches a specific size or time limit.

Real-Life Example

A company starts a new register when the old register becomes full. Similarly, rotating logs create a new log file when the current one becomes too large.

Example

```python
import logging
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler(
    "application.log",
    maxBytes=1000,
    backupCount=3
)

logger = logging.getLogger("MyApp")
logger.setLevel(logging.INFO)
logger.addHandler(handler)

logger.info("Application started")
```

Explanation: When application.log reaches the specified size, older logs are moved to backup files.

### 9. Logging Best Practices

Definition

Logging best practices are recommended ways to use logging effectively and clearly.

Real-Life Example

A company maintains records using proper dates, categories, and clear information instead of writing random notes.

Example

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("User logged in")
logging.warning("Low storage")
logging.error("Database connection failed")
```

```python
Explanation: Good logging should:

Use the correct log level.
Use meaningful messages.
Include useful information such as time and level.
Avoid sensitive information like passwords.
Store logs properly in files when needed.
Use rotating logs for large applications.
```

## Module 5 – File Handling

### 1. Reading Text Files

Definition

Reading a text file means opening a file and retrieving the data stored inside it.

Real-Life Example

Think of opening a notebook and reading what is written inside. Python reads a text file in the same way.

Example

```python
file = open("student.txt", "r")

data = file.read()

print(data)

file.close()
```

Suppose student.txt contains:

Suppose student.txt contains:

Darshan
Python

Output:

Darshan
Python

Explanation: "r" means read mode. read() reads the content from the file.

### 2. Writing Text Files

Definition

Writing a text file means storing data inside a text file.

Real-Life Example

Think of writing information into a notebook. Python can write information into a file in the same way.

Example

```python
file = open("student.txt", "w")

file.write("Darshan is learning Python")

file.close()
```

Content of student.txt:

Darshan is learning Python

Explanation: "w" means write mode. If the file doesn't exist, Python creates it. If it already exists, its old content is replaced.

### 3. CSV Files

Definition

A CSV (Comma-Separated Values) file stores tabular data in rows and columns, usually separated by commas.

Real-Life Example

Think of an Excel sheet containing student names and marks. A CSV file stores similar table-like information.

Example

```python
import csv

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name", "Marks"])
    writer.writerow(["Darshan", 85])
    writer.writerow(["Rahul", 90])
```

Content of students.csv:

Name,Marks
Darshan,85
Rahul,90

Explanation: csv.writer() creates a CSV writer, and writerow() writes one row at a time.

### 4. JSON Files

Definition

JSON (JavaScript Object Notation) is a format used to store and exchange structured data using key-value pairs.

Real-Life Example

Think of a student form:

Name: Darshan
Age: 26
Course: Python

JSON stores information in a similar key-value structure.

```python
import json

student = {
    "name": "Darshan",
    "age": 26,
    "course": "Python"
}

with open("student.json", "w") as file:
    json.dump(student, file)
```

Content of student.json:

{
    "name": "Darshan",
    "age": 26,
    "course": "Python"
}

Explanation: json.dump() converts the Python dictionary and stores it in a JSON file.

### 5. Pickle Files

Definition

Pickle is used to save Python objects into a file and load them back later.

Real-Life Example

Think of packing an object into a box for storage and unpacking it when needed. Pickle does the same with Python objects.

Example

```python
import pickle

student = {
    "name": "Darshan",
    "marks": 85
}

with open("student.pkl", "wb") as file:
    pickle.dump(student, file)

with open("student.pkl", "rb") as file:
    data = pickle.load(file)

print(data)
```

Output:

{'name': 'Darshan', 'marks': 85}

Explanation: pickle.dump() saves the object, while pickle.load() retrieves it.

wb = write binary
rb = read binary

Only load Pickle files from trusted sources because untrusted Pickle data can execute malicious code

### 6. Context Manager (with)

Definition

A context manager using with automatically handles opening and closing resources such as files.

Real-Life Example

Think of an automatic door. You enter, use the room, and the door closes automatically when you leave. Similarly, with automatically closes the file after use.

Example

```python
with open("student.txt", "r") as file:
    data = file.read()
    print(data)
```

Output:

Darshan is learning Python

Explanation: We don't need to use file.close() because with automatically closes the file.

### 7. pathlib

Definition

pathlib is a Python module used to work with file and directory paths in an object-oriented and convenient way.

Real-Life Example

Think of a home address that tells you exactly where a house is located. pathlib helps Python work with file and folder addresses.

Example

```python
from pathlib import Path

file = Path("student.txt")

print(file.exists())
print(file.name)
```

Explanation: Path() represents the file path, exists() checks whether it exists, and name gives the file name.

### 8. os

Definition

The os module is used to interact with the operating system, including files, folders, and paths.

Real-Life Example

Think of File Explorer, where we can create, rename, or remove folders. The os module allows Python to perform similar operations.

Example

```python
import os

os.mkdir("Students")

print(os.path.exists("Students"))
```

Explanation: os.mkdir() creates a directory, and os.path.exists() checks whether it exists.

Definition

The shutil module is used for high-level file operations such as copying, moving, and deleting files or directories.

Real-Life Example

Think of copying a document from one folder to another on your computer. shutil can perform the same operation through Python.

Example

```python
import shutil

shutil.copy("student.txt", "student_backup.txt")
```

Explanation: shutil.copy() creates a copy of the original file

### 10. File Validation

Definition

File validation means checking whether a file is valid before processing it, such as checking whether it exists or has the correct format.

Real-Life Example

Before entering an exam hall, your hall ticket is checked to make sure it is valid. Similarly, we validate a file before using it.

Example

```python
import os

file_path = "student.txt"

if os.path.exists(file_path) and file_path.endswith(".txt"):
    print("Valid file")
else:
    print("Invalid file")
```

Explanation: The program checks whether the file exists and whether it has the .txt extension.

### 11. Directory Operations

Definition

Directory operations are actions performed on folders, such as creating, listing, renaming, and deleting directories.

Real-Life Example

Think of organizing folders on your laptop. You can create a folder, view its contents, rename it, or delete it. Python can do the same.

Example

```python
import os

os.mkdir("PythonFiles")

print(os.listdir("."))

os.rename("PythonFiles", "PythonData")

os.rmdir("PythonData")
```

Explanation: mkdir() creates a directory, listdir() lists contents, rename() changes its name, and rmdir() removes an empty directory.

## Exception Handling

### try

Definition

try is used to write code that may cause an exception. Python first executes the code inside the try block.

Real-Life Example

Imagine trying to open a door.

There is a possibility that the door may be locked or something may go wrong. So you try the action and are prepared to handle the problem.

The try block works similarly—it contains code that might cause an error.

Example

```python
try:
    number = int(input("Enter a number: "))
    print(number)
except:
    print("Invalid input")
```

### except

Definition

except is used to handle an exception that occurs inside the try block.

Real-Life Example

Imagine driving a car and suddenly getting a flat tire.

You don't abandon the car. Instead, you use a spare tire to handle the problem.

except works similarly—it provides a way to handle an error.

Example

```python
try:
    number = int("abc")
except ValueError:
    print("Please enter a valid number")
```

Here, ValueError occurs and the except block handles it.

### else

Definition

else is used to execute code when no exception occurs in the try block.

Real-Life Example

Imagine submitting an online form.

If all the information is correct, the website displays a success message.

The else block works similarly—it runs when the try block executes successfully.

Example

```python
try:
    number = int("10")
except ValueError:
    print("Invalid number")
else:
    print("Number is valid")
```

Here, no exception occurs, so the else block runs.

### finally

Definition

finally is used to execute a block of code whether an exception occurs or not.

Real-Life Example

Imagine borrowing a book from a library.

Whether you successfully read the book or face some problem, you still need to return it.

finally works similarly—it performs an action that should happen at the end.

Example

```python
try:
    number = int("10")
    print(number)
except ValueError:
    print("Invalid number")
finally:
    print("Process completed")
```

### raise

Definition

raise is used to manually generate an exception when a specific condition occurs.

Real-Life Example

Imagine a security guard checking age at an entrance.

If someone is below the required age, the guard stops them and raises an alert.

raise works similarly—it allows the programmer to manually trigger an exception.

Example

```python
age = 15

try:
    if age < 18:
        raise ValueError("Age must be 18 or above")
except ValueError as e:
    print(e)
```

Here, the programmer manually raises a ValueError when the age is below 18

### assert

Definition

assert is used to check whether a condition is True. If the condition is False, Python raises an AssertionError.

Real-Life Example

Imagine a teacher checking that student marks must be between 0 and 100.

If a student's marks are 150, the teacher knows that something is wrong.

assert works similarly—it checks whether an expected condition is True.

Example

```python
marks = 150

try:
    assert marks <= 100, "Marks cannot be greater than 100"
except AssertionError as e:
    print(e)
```

Here, the condition is False, so AssertionError occurs and is handled.

### Custom Exceptions

Definition

A custom exception is an exception created by the programmer for a specific situation in a program.

Real-Life Example

Imagine a bank has a special rule for insufficient balance.

Instead of showing a general error, the bank can show a specific message such as "Insufficient Balance."

Custom exceptions work similarly—they allow us to create our own meaningful exception types.

Example

```python
class InsufficientBalanceError(Exception):
    pass


balance = 1000
withdraw = 1500

try:
    if withdraw > balance:
        raise InsufficientBalanceError("Insufficient balance")
except InsufficientBalanceError as e:
    print(e)
```

Here, InsufficientBalanceError is a custom exception created by us.

### Built-in Exceptions

ValueError
Definition

ValueError occurs when a function receives a value with the correct data type but an inappropriate value.

Real-Life Example

Imagine a form asking for your age, and you enter "twenty" when the system expects a number.

The value cannot be converted into an integer.

Example

```python
try:
    age = int("twenty")
except ValueError:
    print("Invalid age")
```

Here, "twenty" cannot be converted into an integer, so ValueError occurs.

### TypeError

Definition

TypeError occurs when an operation is performed using an inappropriate data type.

Real-Life Example

Imagine trying to add a number and a person's name.

For example, adding 10 and "Darshan" does not make sense because they are different types of data.

Example

```python
try:
    result = 10 + "20"
except TypeError:
    print("Cannot add integer and string")
```

Here, Python cannot add an integer and a string, so TypeError occurs.

### IndexError

Definition

IndexError occurs when you try to access an index that does not exist in a sequence such as a list or tuple.

Real-Life Example

Imagine a classroom has 3 students.

If you ask for the 10th student from the list, that student does not exist.

Example

```python
students = ["Darshan", "Rahul", "Arun"]

try:
    print(students[5])
except IndexError:
    print("Index does not exist")
```

### KeyError

Definition

KeyError occurs when you try to access a key that does not exist in a dictionary.

Real-Life Example

Imagine looking for a student's details using their student ID.

If that ID is not present in the college records, the information cannot be found.

Example

```python
student = {
    "name": "Darshan",
    "age": 24
}

try:
    print(student["marks"])
except KeyError:
    print("Key does not exist")
```

Here, "marks" is not present in the dictionary, so KeyError occurs.

### AttributeError

Definition

AttributeError occurs when you try to access an attribute or method that an object does not have.

Real-Life Example

Imagine a car has an accelerator and steering wheel.

If you try to use a feature that the car does not have, the action cannot be performed.

Example

```python
name = "Darshan"

try:
    name.fly()
except AttributeError:
    print("Object does not have this method")
```

Here, a string does not have a fly() method, so AttributeError occurs.

### NameError

Definition

NameError occurs when you try to use a variable or name that has not been defined.

Real-Life Example

Imagine a teacher calls the name of a student who is not registered in the class.

The student cannot be found.

NameError works similarly when Python cannot find a defined name.

Example

```python
try:
    print(student_name)
except NameError:
    print("Variable is not defined")
```

Here, student_name was never defined, so NameError occurs.

### FileNotFoundError

Definition

FileNotFoundError occurs when Python tries to access a file that does not exist at the specified location.

Real-Life Example

Imagine going to a library shelf to find a book, but the book is not there.

The file system works similarly when the requested file cannot be found.

Example

```python
try:
    file = open("student_data.txt", "r")
except FileNotFoundError:
    print("File not found")
```

Here, if student_data.txt does not exist, FileNotFoundError occurs.

```python
### ZeroDivisionError

Definition

ZeroDivisionError occurs when you try to divide a number by zero.

Real-Life Example

Imagine trying to share 100 chocolates among zero people.

There is nobody to receive the chocolates, so the operation cannot be performed.

Example
```

```python
try:
    result = 100 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
```

Here, dividing 100 by 0 causes ZeroDivisionError.

### ImportError

Definition

ImportError occurs when Python cannot import a required name or item from a module.

Real-Life Example

Imagine going to a shop that exists and asking for a product that the shop does not have.

The shop exists, but the requested item is unavailable.

Example

```python
try:
    from math import addition
except ImportError:
    print("Cannot import addition from math")
```

### ModuleNotFoundError

Definition

ModuleNotFoundError occurs when Python cannot find the module that you are trying to import.

Real-Life Example

Imagine going to a shop that does not exist.

You cannot buy anything because the shop itself cannot be found.

ModuleNotFoundError works similarly when Python cannot find the requested module.

Example

```python
try:
    import my_unknown_module
except ModuleNotFoundError:
    print("Module not found")
```

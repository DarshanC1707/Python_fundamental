## Module 2 – Object-Oriented Programming

### Class

Definition

A class is a blueprint or template used to create objects. It defines the properties and actions that the objects can have.

Real-Life Example

Imagine a house blueprint.

The blueprint defines where the rooms, doors, windows, and kitchen should be. But the blueprint itself is not an actual house.

A class works the same way—it defines what an object should contain and do.

Example

```python
class Student:
    name = "Darshan"
    age = 24


student1 = Student()

print(student1.name)
print(student1.age)
```

Here, Student is the class. It defines the properties name and age.

### Object

Definition

An object is an instance of a class. It is the actual entity created using the class.

Real-Life Example

Imagine the house blueprint is the class, and the actual houses built using that blueprint are objects.

One blueprint can be used to build many houses.

Similarly, one class can be used to create multiple objects.

Example

```python
class Student:
    name = "Darshan"

student1 = Student()
student2 = Student()

print(student1.name)
print(student2.name)
```

### Constructor __init__()

Definition

A constructor is a special method that is automatically called when an object is created. In Python, the constructor is written as __init__().

Real-Life Example

Imagine ordering a new school ID card.

When the card is created, the student's name, ID number, and department are entered automatically.

The __init__() method works similarly—it initializes the object's information when the object is created.

Example

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Darshan", 24)

print(student1.name)
print(student1.age)
```

Here, __init__() runs automatically when student1 is created and assigns the values to the object.

### Instance Variables

Definition

Instance variables are variables that belong to a particular object. Each object can have its own values.

Real-Life Example

Imagine a classroom with many students.

Each student has their own name and marks. One student's marks can be different from another student's marks.

Instance variables work the same way—each object can have its own data.

Example

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student1 = Student("Darshan", 80)
student2 = Student("Rahul", 90)

print(student1.name, student1.marks)
print(student2.name, student2.marks)
```

Here, name and marks are instance variables because each student object has its own values.

Real-Life Example

Imagine all students in a college belong to the same college name.

A class variable works the same way—it stores common information shared by all objects.

### Class Variables

Definition

Class variables are variables that are shared by all objects of a class.

Real-Life Example

Imagine all students in a college belong to the same college name.

The college name is common to every student, so we don't need to store it separately for every student.

A class variable works the same way—it stores common information shared by all objects.

Example

```python
class Student:

    college = "University of Mysore"

    def __init__(self, name):
        self.name = name

student1 = Student("Darshan")
student2 = Student("Rahul")

print(student1.college)
print(student2.college)
```

Here, college is a class variable because it is common to all student objects.

### Instance Methods

Definition

Instance methods are methods that work with the data of a particular object. They use self to access instance variables.

Real-Life Example

Imagine each student can introduce themselves.

Darshan introduces himself with his own name, while Rahul introduces himself with his own name.

The method works with the data of the particular student.

Example

```python
class Student:

    def __init__(self, name):
        self.name = name

    def introduce(self):
        print("My name is", self.name)

student1 = Student("Darshan")
student2 = Student("Rahul")

student1.introduce()
student2.introduce()
```

Here, introduce() is an instance method because it uses self.name.

### Static Methods

Definition

A static method is a method that belongs to a class but does not depend on the class or object data. It is created using @staticmethod.

Real-Life Example

Imagine a school has a general rule to check whether a number is even or odd.

This calculation does not depend on any particular student.

A static method works similarly—it performs a general task without using object-specific data.

Example

```python
class Student:

    @staticmethod
    def check_even(number):
        return number % 2 == 0

print(Student.check_even(10))
```

```python
Here, check_even() does not use self or any object data, so it can be a static method.
```

### Class Methods

Definition

A class method is a method that works with the class itself rather than a particular object. It is created using @classmethod and uses cls.

Real-Life Example

Imagine a college wants to change its college name for all students.

Instead of changing it separately for every student, the college-level information can be changed once.

A class method works similarly—it can work with class-level data.

Example

```python
class Student:

    college = "ABC College"

    @classmethod
    def change_college(cls, name):
        cls.college = name

Student.change_college("XYZ College")

print(Student.college)
```

Here, cls refers to the class itself, and the class variable college is changed.

### Properties @property

Definition

A property allows a method to be accessed like a normal variable. It is created using the @property decorator.

Real-Life Example

Imagine a bank account.

You can ask the bank for your current balance like you are simply checking a value. You don't need to know how the bank calculates or retrieves that balance internally.

@property works similarly—it allows us to access calculated or controlled data like an attribute.

Example

```python
class Student:

    def __init__(self, name):
        self.name = name

    @property
    def student_name(self):
        return self.name

student1 = Student("Darshan")


print(student1.student_name)
```

### Encapsulation

Definition

Encapsulation is the concept of keeping data and the methods that operate on that data together inside a class and controlling how the data is accessed.

Real-Life Example

Imagine an ATM machine.

You can withdraw money or check your balance using the available options, but you cannot directly access the bank's internal database.

Encapsulation works similarly—it protects internal data and provides controlled ways to access or modify it.

Example

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)

print(account.get_balance())
```

Here, __balance is treated as private data. We access it through the get_balance() method instead of directly accessing it.

### Abstraction

Definition

Abstraction is the concept of hiding unnecessary internal details and showing only the important functionality to the user.

Real-Life Example

Imagine driving a car.

You simply use the steering wheel, accelerator, and brake to drive the car. You don't need to know how the engine, fuel injection, or transmission works internally.

Abstraction works the same way—it hides the complex implementation and provides only the required functionality.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car started")


car = Car()

car.start()
```

### Inheritance

Definition

Inheritance is an OOP concept where one class can use the properties and methods of another class.

The existing class is called the parent class, and the new class is called the child class.

Real-Life Example

Imagine a parent teaching skills to their child.

The child can learn common skills from the parent and can also have their own additional skills.

Inheritance works the same way—a child class can reuse the features of a parent class.

Example

```python
class Animal:

    def eat(self):
        print("Animal is eating")


class Dog(Animal):

    def bark(self):
        print("Dog is barking")


dog = Dog()

dog.eat()
dog.bark()
```

Here, Dog inherits the eat() method from Animal and also has its own bark() method.

### Multiple Inheritance

Definition

Multiple inheritance is a type of inheritance where one child class inherits from two or more parent classes.

Real-Life Example

Imagine a person learning cooking from one parent and programming from another parent.

The person can have skills from both sources.

Multiple inheritance works similarly—a child class can inherit features from multiple parent classes.

Example

```python
class Father:

    def driving(self):
        print("Can drive")


class Mother:

    def cooking(self):
        print("Can cook")


class Child(Father, Mother):
    pass


child = Child()

child.driving()
child.cooking()
```

Here, Child inherits from both Father and Mother.

### Multilevel Inheritance

Definition

Multilevel inheritance is a type of inheritance where a class inherits from another class, which itself inherits from another class.

Real-Life Example

Imagine a skill being passed through three generations.

A grandparent teaches something to a parent, and the parent passes it to the child.

Multilevel inheritance works similarly—the child receives features through multiple levels of inheritance.

Example

```python
class Grandfather:

    def house(self):
        print("Grandfather's house")


class Father(Grandfather):

    def car(self):
        print("Father's car")


class Son(Father):

    def bike(self):
        print("Son's bike")


son = Son()

son.house()
son.car()
son.bike()
```

Here, Son inherits from Father, and Father inherits from Grandfather.

### Hierarchical Inheritance

Definition

Hierarchical inheritance is a type of inheritance where multiple child classes inherit from the same parent class.

Real-Life Example

Imagine a school teacher teaching the same basic subject to many students.

All students receive the common knowledge from the same teacher, but each student can also have their own skills.

Hierarchical inheritance works similarly—multiple child classes share one parent class.

Example

```python
class Animal:

    def eat(self):
        print("Animal is eating")


class Dog(Animal):

    def bark(self):
        print("Dog is barking")


class Cat(Animal):

    def meow(self):
        print("Cat is meowing")


dog = Dog()
cat = Cat()

dog.eat()
dog.bark()

cat.eat()
cat.meow()
```

Here, both Dog and Cat inherit from the same Animal class.

### Polymorphism

Definition

Polymorphism means one interface or method name can have different behavior depending on the object using it.

Real-Life Example

Imagine the command "speak".

A dog responds by barking, while a cat responds by meowing.

The same action—speak()—produces different results depending on the animal.

Polymorphism works the same way.

Example

```python
class Dog:

    def speak(self):
        print("Dog barks")


class Cat:

    def speak(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.speak()
cat.speak()
```

Here, both classes have the same method speak(), but each object behaves differently.

### Method Overriding

Definition

Method overriding occurs when a child class provides its own version of a method that already exists in the parent class.

Real-Life Example

Imagine a parent says "I drive a car."

The child learns the same concept but says "I drive a bike."

The child replaces the parent's version with its own behavior.

Method overriding works similarly.

Example

```python
class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Dog barks")


dog = Dog()

dog.sound()
```

```python
Here, Dog overrides the sound() method of Animal.
```

### Method Overloading (Python Approach)

Definition

Method overloading means using the same method name with different numbers or types of arguments.

Python does not support traditional method overloading like some other programming languages. Instead, we can use default arguments or *args.

Real-Life Example

Imagine a restaurant's order system.

You can place an order with one item, two items, or several items using the same ordering process.

The method can handle different numbers of inputs.

Example

```python
class Calculator:

    def add(self, a, b=0, c=0):
        return a + b + c


calculator = Calculator()

print(calculator.add(10))
print(calculator.add(10, 20))
print(calculator.add(10, 20, 30))
```

Here, the same add() method works with different numbers of arguments using default values.

### Composition

Definition

Composition is an OOP relationship where one class contains an object of another class, and the contained object is strongly dependent on the main object.

Real-Life Example

Imagine a house and its rooms.

A house contains rooms. If the house is destroyed, those rooms as part of that house no longer exist.

Composition represents this strong has-a relationship.

Example

```python
class Engine:

    def start(self):
        print("Engine started")


class Car:

    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
        print("Car started")


car = Car()

car.start_car()
```

### Aggregation

Definition

Aggregation is an OOP relationship where one class contains an object of another class, but the contained object can exist independently.

Real-Life Example

Imagine a college and students.

A college has students, but a student can continue to exist even if they leave that particular college.

This is a weaker has-a relationship.

Example

```python
class Student:

    def __init__(self, name):
        self.name = name


class College:

    def __init__(self, student):
        self.student = student


student = Student("Darshan")

college = College(student)

print(college.student.name)
```

Here, the Student object is created separately and then given to the College.

### Association

Definition

Association is a relationship where two independent classes interact with each other.

Real-Life Example

Imagine a teacher and student.

A teacher teaches a student, but both can exist independently.

They simply interact with each other.

Example

```python
class Teacher:

    def teach(self):
        print("Teacher is teaching")


class Student:

    def learn(self, teacher):
        teacher.teach()
        print("Student is learning")


teacher = Teacher()
student = Student()

student.learn(teacher)
```

Here, Teacher and Student are independent objects that interact with each other.

### Magic (Dunder) Methods

Definition

Magic methods, also called dunder methods, are special methods in Python that have double underscores before and after their names.

They allow Python objects to work with built-in operations such as printing, length checking, comparison, and more.

Real-Life Example

Imagine a smartphone.

When you press the power button, the phone automatically performs several internal actions.

You don't manually tell every internal component what to do.

Magic methods work similarly—Python automatically calls certain special methods when specific operations are performed.

Example

```python
class Student:

    def __str__(self):
        return "Student object"


student = Student()

print(student)
```

Here, Python automatically calls the __str__() magic method when we use print(student).

### super()

Definition

super() is used to access methods or variables from the parent class inside a child class.

Real-Life Example

Imagine a child wants to use something that was already taught by their parent, and then add something of their own.

Instead of writing the parent's work again, the child can use it directly.

super() works similarly—it allows the child class to use the parent's functionality.

Example

```python
class Parent:

    def show(self):
        print("Parent method")


class Child(Parent):

    def show(self):
        super().show()
        print("Child method")


child = Child()

child.show()
```

### __str__()

Definition

__str__() is a magic method used to define the readable string representation of an object.

Real-Life Example

Imagine a student ID card.

When someone looks at it, they see useful information such as the student's name instead of some internal system details.

__str__() works similarly—it defines what should be displayed when we print an object.

Example

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student Name: {self.name}"


student = Student("Darshan")

print(student)
```

Here, __str__() controls what is displayed when the object is printed.

### __repr__()

Definition

__repr__() is a magic method used to provide a detailed or developer-friendly representation of an object.

Real-Life Example

Imagine a product in an inventory system.

A customer may only need to see the product name, but a developer may need detailed information such as the product ID and price.

__repr__() is useful for this detailed representation.

Example

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __repr__(self):
        return f"Student(name='{self.name}', marks={self.marks})"


student = Student("Darshan", 85)

print(repr(student))
```

Here, __repr__() provides a detailed representation of the object.

### __len__()

Definition

__len__() is a magic method used to define what should happen when the len() function is used with an object.

Real-Life Example

Imagine a classroom.

When someone asks, "How many students are in this classroom?", the classroom provides the number of students.

__len__() works similarly—it tells Python how to calculate the length of an object.

Example

```python
class Classroom:

    def __init__(self, students):
        self.students = students

    def __len__(self):
        return len(self.students)


classroom = Classroom(["Darshan", "Rahul", "Arun"])

print(len(classroom))
```

Here, __len__() tells Python to return the number of students.

### Dataclasses (Bonus)

Definition

A dataclass is a special type of Python class that automatically provides common methods for storing and working with data.

Real-Life Example

Imagine maintaining student records.

For every student, you need to store information such as name and marks.

Instead of writing the constructor and other basic methods manually every time, a dataclass can automatically create much of this code.

Example

```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    marks: int


student = Student("Darshan", 85)

print(student)
```

Here, @dataclass automatically creates useful methods such as __init__() and __repr__().

### Singleton Pattern (Bonus)

Definition

The Singleton Pattern is a design pattern that ensures only one object of a class is created and reused throughout the program.

Real-Life Example

Imagine a school has one main principal's office.

Everyone who needs the principal's office uses the same office instead of creating a new office every time.

A Singleton works similarly—only one instance of the class is created.

Example

```python
class Singleton:

    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance


obj1 = Singleton()
obj2 = Singleton()

print(obj1 is obj2)
```

Here, obj1 and obj2 refer to the same object, so only one instance of Singleton is created.

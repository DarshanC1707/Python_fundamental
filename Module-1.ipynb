## Control Flow

### 1. if / elif / else

if, elif, and else are used to check conditions and decide which block of code should be executed. We can use them when the result needs to change based on a particular condition.

if → checks the first condition
elif → checks another condition when the previous condition is false
else → executes when none of the conditions are true


```python
#Example:

marks = 82

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
else:
    grade = "C"

print("Grade:", grade)
```

Here, marks is 82. The first condition marks >= 90 is false, so Python checks the elif condition. Since 82 >= 75 is true, the grade is assigned as "B".

```python
#Another example:

age = 20

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

### 2. for Loop

A for loop is used when we want to repeat the same block of code for each item in a sequence.

```python
# Example

for number in range(1, 6):
    print(number)
```

Here, range(1, 6) provides the numbers from 1 to 5, and the for loop processes each number one by one.

So, a for loop helps us repeat an operation for each item in a sequence or range of values.

### 3. while Loop

A while loop repeats a block of code as long as a given condition remains true. The condition is checked before each repetition.

```python
#Example:

count = 1

while count <= 5:
    print(count)
    count += 1
```

Here, count starts with 1. The loop continues as long as count <= 5 is true. After each repetition, count is increased by 1.

When count becomes 6, the condition becomes false and the loop stops.

### 4. break

break is used to stop a loop immediately, even if the loop could continue further.

```python
#Example:

for number in range(1, 6):
    if number == 3:
        break
    print(number)
```

Here, the loop starts with numbers from 1 to 5. When number becomes 3, the if condition becomes true and break stops the loop immediately.

### 5. continue

continue is used to skip the current repetition of a loop and move to the next iteration

```python
#Example:

for number in range(1, 6):
    if number == 3:
        continue
    print(number)
```

Here, when number becomes 3, continue skips the print() statement for that iteration.

The loop does not stop. It continues with the next number.

“Here, continue skips the current iteration when the number is 3, so 3 is not printed. The loop then continues with the next number.”

### 6. pass

pass is used when we need a statement syntactically, but we do not want to perform any action at that point.

It acts as a placeholder and allows the code to continue without doing anything.

```python
#Example:

for number in range(1, 6):
    if number == 3:
        pass
    else:
        print(number)
```

Here, when number is 3, the pass statement does nothing. The program continues with the next part of the code.

pass means do nothing.

Because the if condition was true, the else block is not executed.

### 5. List Comprehension

List comprehension provides a shorter way to create a list by performing an operation for each item.

Instead of writing a complete for loop and creating the list separately, we can create the list in a single expression.

```python
numbers = [1, 2, 3, 4, 5]

l = []

for i in numbers:
    l.append(i + 10)

print(l)
```

We can write the same thing using list comprehension:

```python
numbers = [1, 2, 3, 4, 5]

l = [i + 10 for i in numbers]

print(l)
```

So, list comprehension is a compact way to create a new list from an existing collection.

### 6. Dictionary Comprehension

```python
numbers = [1, 2, 3, 4, 5]

l = {number: number * 10 for number in numbers}

print(l)
```

Take each number from numbers.

Use number as the key.
    
Use number * 10 as the value.
    
Store them as key-value pairs.

### 2. Generators

A generator is a special type of function that generates values one at a time instead of creating all the values in memory at once.

```python
def numbers():
    yield 10
    yield 20
    yield 30

for number in numbers():
    print(number)
```

The important thing is that the generator doesn't need to store all values at once.

For example, imagine generating numbers from 1 to 10000. Creating a list would require a lot of memory.

A generator can produce:

1 → process
2 → process
3 → process
...

one at a time.


Why do we use generators?

Mainly for memory efficiency and processing large datasets.

Simple definition

A generator is a function that produces values one at a time using yield, which helps save memory.

### 4. Decorators

A decorator allows us to add extra functionality to an existing function without modifying the original function's code.

```python
#Example:

def logger(func):
    def wrapper():
        print("Function started")
        func()
        print("Function finished")

    return wrapper


@logger
def greet():
    print("Hello")


greet()
```

Here, greet() originally only prints:

Hello

The decorator adds:

Function started

Function finished

without changing the greet() function itself.

### 5. Closures

A closure occurs when an inner function remembers a variable from its outer function, even after the outer function has finished executing

```python
def outer():
    name = "Dog"

    def inner():
        print(name)

    return inner


my_function = outer()

my_function()
```

## Built-in Functions
### map()

Definition

map() is a built-in function used to apply the same operation or function to every item in a sequence.

Real-Life Example

Imagine a teacher has marks of 5 students and wants to add 5 grace marks to every student's marks.

Instead of adding 5 separately to each student's marks, the teacher applies the same rule to everyone.

map() works the same way—it applies the same function to every item.

```python
marks = [50, 60, 70, 80, 90]

def add_grace(mark):
    return mark + 5

result = map(add_grace, marks)

print(list(result))
```

What map() does internally is basically:

50 → add_grace(50) → 55

60 → add_grace(60) → 65

70 → add_grace(70) → 75

80 → add_grace(80) → 85

90 → add_grace(90) → 95

So:

map(add_grace, marks)

means:

"Apply add_grace() to every item in marks."

### filter()

Definition

filter() is a built-in function used to select items from a sequence based on a condition.

Real-Life Example

Imagine a teacher wants to find only the students who scored more than 60 marks.

The teacher checks each student's marks and keeps only the students who satisfy the condition.

filter() works the same way—it keeps only the items that satisfy a condition.

```python
marks = [45, 60, 75, 50, 90]

def above_60(mark):
    return mark > 60

result = filter(above_60, marks)

print(list(result))
```

### zip()

Definition

zip() is a built-in function used to combine two or more sequences by matching their items based on position.

Real-Life Example

Imagine a teacher has two lists:

Student marks

The teacher wants to match each student with their corresponding mark.

zip() works the same way—it combines items from different sequences based on their position.

```python
names = ["Darshan", "Rahul", "Arun"]
marks = [80, 75, 90]

result = zip(names, marks)

print(list(result))
```

Here, zip() matches the first name with the first mark, second name with second mark, and so on.

### enumerate()

Definition

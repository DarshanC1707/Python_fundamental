### 2. Tables

Definition

A table stores related data in the form of rows and columns inside a database.

Real-Life Example

Think of an Excel sheet where each row represents a student and each column represents information such as name, age, and marks.

Example

```python
CREATE TABLE students (
    student_id INT,
    name VARCHAR(50),
    marks INT
);
```

### 3. Primary Key

Definition

A Primary Key is a column that uniquely identifies each row in a table. It cannot contain duplicate or NULL values.

Real-Life Example

A student ID uniquely identifies each student in a college.

Example

```python
CREATE TABLE students (
    student_id INT PRIMARY KEY,
    name VARCHAR(50),
    marks INT
);
```

```python
Example Data:

1 | Darshan | 85
2 | Rahul   | 90

Explanation: student_id uniquely identifies every student.
```

### 4. Foreign Key

Definition

A Foreign Key is a column that connects one table to the primary key of another table.

Real-Life Example

A student's department_id can connect the student to their department.

Example

```python
CREATE TABLE departments (
    department_id INT PRIMARY KEY,
    department_name VARCHAR(50)
);

CREATE TABLE students (
    student_id INT PRIMARY KEY,
    name VARCHAR(50),
    department_id INT,
    FOREIGN KEY (department_id)
        REFERENCES departments(department_id)
);
```

```python
Explanation: department_id connects the students table with the departments table.
```

Definition

Constraints are rules applied to columns to control what data can be stored.

Common constraints include:

PRIMARY KEY, FOREIGN KEY, NOT NULL, UNIQUE, CHECK, and DEFAULT.

## Queries

### 7. SELECT
Definition

SELECT is used to retrieve data from a table.

Real-Life Example

Think of asking the college system to show all student names and marks.

Example

```python
SELECT name, marks
FROM students;
```

```python
Example Output:

Darshan | 85
Rahul   | 90
Priya   | 75

Explanation: SELECT retrieves the requested columns.
```

### 8. WHERE

Definition

WHERE is used to filter rows based on a condition.

Real-Life Example

Suppose a teacher wants only students who scored more than 80 marks.

Example

```python
SELECT *
FROM students
WHERE marks > 80;
```

```python
Example Output:

1 | Darshan | 85
2 | Rahul   | 90

Explanation: Only students with marks greater than 80 are returned.
```

### 9. ORDER BY

Definition

ORDER BY is used to sort query results in ascending or descending order.

Real-Life Example

A teacher arranges students from highest marks to lowest marks.

Example

```python
SELECT name, marks
FROM students
ORDER BY marks DESC;
```

```python
Example Output:

Rahul   | 90
Darshan | 85
Priya   | 75

Explanation: DESC sorts marks from highest to lowest. ASC sorts from lowest to highest.
```

### 10. GROUP BY

Definition

GROUP BY groups rows that have the same value so aggregate functions can be applied.

Real-Life Example

A college wants to find the number of students in each department.

Example

```python
SELECT department_id, COUNT(*) AS total_students
FROM students
GROUP BY department_id;
```

```python
Example Output:

101 | 3
102 | 2

Explanation: Students are grouped according to their department.
```

### 11. HAVING

Definition

HAVING is used to filter grouped results after GROUP BY.

Real-Life Example

A college wants to display only departments having more than two students.

Example

```python
SELECT department_id, COUNT(*) AS total_students
FROM students
GROUP BY department_id
HAVING COUNT(*) > 2;
```

```python
Example Output:

101 | 3

Explanation: HAVING filters the groups after they are created.
```

### 12. DISTINCT

Definition

DISTINCT removes duplicate values from query results.

Real-Life Example

If many students belong to the same department, we may want to see each department only once.

Example

```python
SELECT DISTINCT department_id
FROM students;

Example Output:

101
102
103

Explanation: Duplicate department IDs are removed.
```

### 13. LIMIT

Example

### 14. INNER JOIN

Definition

INNER JOIN returns only the records that have matching values in both tables.

Real-Life Example

Show only students who are assigned to an existing department.

```python
SELECT students.name, departments.department_name
FROM students
INNER JOIN departments
ON students.department_id = departments.department_id;
```

```python
Output:

Darshan | CSE
Rahul   | ECE

Explanation: Only matching records from both tables are returned.
```

### 15. LEFT JOIN

Definition

LEFT JOIN returns all rows from the left table and matching rows from the right table.

Real-Life Example

Show every student even if they haven't been assigned a department.

Example

```python
SELECT students.name, departments.department_name
FROM students
LEFT JOIN departments
ON students.department_id = departments.department_id;
```

```python
Output:

Darshan | CSE
Rahul   | ECE
Priya   | NULL

Explanation: All students are returned. Missing dep artment information becomes NULL.
```

### 16. RIGHT JOIN

Definition

RIGHT JOIN returns all rows from the right table and matching rows from the left table.

Real-Life Example

Show every department even if it currently has no students.

Example

```python
SELECT students.name, departments.department_name
FROM students
RIGHT JOIN departments
ON students.department_id = departments.department_id;
```

```python
Output:

Darshan | CSE
Rahul   | ECE
NULL    | MECH

Explanation: All departments are returned, including departments without students.
```

### 17. FULL JOIN

Definition

FULL JOIN returns all records from both tables, whether they match or not.

Real-Life Example

Show all students and all departments, including students without departments and departments without students.

Example

```python
SELECT students.name, departments.department_name
FROM students
FULL OUTER JOIN departments
ON students.department_id = departments.department_id;
```

```python
Output:

Darshan | CSE
Rahul   | ECE
Priya   | NULL
NULL    | MECH

Explanation: Both matching and non-matching records are returned.

Note: Some databases such as MySQL do not directly support FULL OUTER JOIN.
```

### 18. CROSS JOIN

Definition

CROSS JOIN returns every possible combination of rows from two tables.

Real-Life Example

If there are 3 students and 2 courses, each student is paired with every course.

Example

```python
SELECT students.name, courses.course_name
FROM students
CROSS JOIN courses;
```

```python
Output:

Darshan | Python
Darshan | SQL
Rahul   | Python
Rahul   | SQL

Explanation: Every student is combined with every course.
```

## Advanced SQL

### 19. UNION

Definition

UNION combines the results of two or more SELECT queries and removes duplicates.

Real-Life Example

Combining student names from two different classes into one list.

Example

```python
SELECT name FROM class_a
UNION
SELECT name FROM class_b;
```

```python
Output:

Darshan
Rahul
Priya

Explanation: Results from both queries are combined into one result.
```

### 20. CASE

Definition

CASE adds conditional logic inside an SQL query.

Real-Life Example

A teacher categorizes students as Pass or Fail depending on their marks.

Example

```python
SELECT name, marks,
CASE
    WHEN marks >= 40 THEN 'Pass'
    ELSE 'Fail'
END AS result
FROM students;
```

```python
Output:

Darshan | 85 | Pass
Rahul   | 90 | Pass
Arun    | 35 | Fail

Explanation: CASE checks the condition and returns the appropriate value.
```

### 21. Subqueries

Definition

A subquery is a query written inside another SQL query.

Real-Life Example

A teacher first calculates the class average and then finds students who scored above that average.

Example

```python
SELECT name, marks
FROM students
WHERE marks > (
    SELECT AVG(marks)
    FROM students
);
```

```python
Output:

Darshan | 85
Rahul   | 90

Explanation: The inner query calculates the average first. The outer query then finds students above that average.
```

### 22. Common Table Expressions (CTEs)

Definition

A CTE is a temporary named result set created using WITH that can be used inside a query.

Real-Life Example

Think of making a temporary shortlist of high-scoring students and then using that shortlist for further work.

Example

```python
WITH high_scores AS (
    SELECT name, marks
    FROM students
    WHERE marks > 80
)

SELECT *
FROM high_scores;
```

```python
Output:

Darshan | 85
Rahul   | 90

Explanation: high_scores temporarily stores the result and is then used by the main query.
```

### 23. Views

Definition

A View is a saved SQL query that behaves like a virtual table.

Real-Life Example

Think of creating a shortcut to frequently viewed student results instead of writing the same query every time.

Example

```python
CREATE VIEW top_students AS
SELECT name, marks
FROM students
WHERE marks > 80;
```

```python
Then:

SELECT *
FROM top_students;
```

```python
Output:

Darshan | 85
Rahul   | 90

Explanation: The view saves the query definition so it can be reused.
```

```python
SELECT name, marks,
ROW_NUMBER() OVER (ORDER BY marks DESC) AS row_num
FROM students;
```

```python
Output:

Rahul   | 90 | 1
Darshan | 85 | 2
Priya   | 75 | 3

Explanation: Every row receives a unique sequential number.
```

### 26. RANK()

Definition

RANK() assigns ranks based on values. Equal values receive the same rank, and the next rank is skipped.

Real-Life Example

If two students share 1st place, the next student gets 3rd place.

Example

```python
SELECT name, marks,
RANK() OVER (ORDER BY marks DESC) AS student_rank
FROM students;
```

```python
Output:

Darshan | 90 | 1
Rahul   | 90 | 1
Priya   | 80 | 3

Explanation: Darshan and Rahul share rank 1, so rank 2 is skipped.
```

### 27. DENSE_RANK()

Definition

DENSE_RANK() assigns the same rank to equal values but does not skip the next rank.

Real-Life Example

If two students share position 1, the next student is given position 2.

Example

```python
SELECT name, marks,
DENSE_RANK() OVER (ORDER BY marks DESC) AS student_rank
FROM students;
```

```python
Output:

Darshan | 90 | 1
Rahul   | 90 | 1
Priya   | 80 | 2

Explanation: Equal marks receive the same rank, but no rank number is skipped.
```

### 28. LAG()

Definition

LAG() accesses a value from the previous row.

Real-Life Example

While checking today's sales, you also look at yesterday's sales for comparison.

Example

```python
SELECT name, marks,
LAG(marks) OVER (ORDER BY student_id) AS previous_marks
FROM students;
```

```python
Output:

Darshan | 85 | NULL
Rahul   | 90 | 85
Priya   | 75 | 90

Explanation: LAG() returns the marks from the previous row.
```

### 29. LEAD()

Definition

LEAD() accesses a value from the next row.

Real-Life Example

While checking today's schedule, you look at what task is coming next.

Example

```python
SELECT name, marks,
LEAD(marks) OVER (ORDER BY student_id) AS next_marks
FROM students;
```

```python
Output:

Darshan | 85 | 90
Rahul   | 90 | 75
Priya   | 75 | NULL

Explanation: LEAD() returns the value from the next row.
```
